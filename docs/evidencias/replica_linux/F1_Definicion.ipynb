## Problema y contexto

La comparación de precios de alojamientos de corta estancia requiere distinguir ofertas con características y condiciones de reserva diferentes. Una tarifa anunciada para un alojamiento completo, una determinada capacidad y ciertas fechas de estancia no representa necesariamente una referencia comparable para otro tipo de oferta.

El proyecto estudiará los anuncios de Airbnb de Buenos Aires disponibles en el archivo detallado de Inside Airbnb identificado en su ruta como `2026-09-23`. La copia revisada contiene 29.753 anuncios y 90 variables. El validador del curso informó cumplimiento de sus requisitos mínimos, con un aviso por valores faltantes. La [ficha de procedencia](../docs/datos/fuente_dataset.md) y el [diccionario de variables](../docs/datos/diccionario_variables.md) documentan las características y las limitaciones de esta fuente.

El problema de análisis consiste en establecer referencias descriptivas de precio por barrio, tipo de alojamiento y capacidad, considerando que las cotizaciones del archivo corresponden a distintas fechas y duraciones de estancia. Las diferencias observadas pueden reflejar tanto las características de los anuncios como la composición de los grupos y las condiciones cotizadas.

(Inside Airbnb, 2026).

## Pregunta y objetivos

### Pregunta

**¿Cómo varía el precio cotizado por noche en ARS según el barrio, el tipo de alojamiento y la capacidad de los anuncios de Airbnb de Buenos Aires, considerando la fecha y la duración de la estancia cotizada?**

La pregunta se refiere a diferencias descriptivas entre los anuncios incluidos en la extracción. No plantea identificar efectos causales ni estimar el precio de una reserva efectivamente realizada.

### Objetivo general

**Caracterizar las diferencias del precio cotizado por noche en ARS entre barrios, tipos de alojamiento y capacidades de los anuncios de Airbnb de Buenos Aires, mediante un análisis descriptivo reproducible que considere la fecha y la duración de las estancias cotizadas.**

### Objetivos específicos

1. **Evaluar la calidad y la cobertura de las variables necesarias para el análisis**, verificando identificadores, faltantes, formatos, moneda y coherencia de las fechas y los importes de las cotizaciones.
2. **Preparar una base analítica trazable**, mediante reglas justificadas de selección, conversión y tratamiento de registros, documentando sus efectos sobre el número de anuncios y la composición de los grupos.
3. **Describir la distribución del precio cotizado por noche según barrio, tipo de alojamiento y capacidad**, mediante el tamaño de cada grupo, la mediana, los cuartiles y representaciones gráficas pertinentes.
4. **Evaluar la estabilidad de las diferencias descriptivas al considerar condiciones de estancia comparables y decisiones de preprocesamiento**, comunicando la cobertura, las limitaciones y el alcance de las referencias obtenidas.

## Supuestos

- Un `id` identifica un anuncio dentro del archivo; su ausencia o repetición se comprobará antes de usarlo para trazar registros.
- La moneda ARS se reconocerá mediante `quote.currency`, sin deducirla únicamente del símbolo monetario. Los importes sin moneda comprobable no se equipararán automáticamente con los demás.
- Las fechas de entrada y salida describen la estancia solicitada para una cotización. No acreditan que esa estancia se haya reservado o realizado.
- Las categorías de barrio y tipo de alojamiento se interpretan según la clasificación de la fuente; su significado y cobertura se revisarán antes de comparar grupos.
- La falta de información no representa un cero. Tampoco se presupone que los anuncios con precio informado tengan la misma composición que los anuncios sin precio.
- Las comparaciones describirán anuncios observados en condiciones documentadas. No se presupone representatividad del mercado ni independencia entre anuncios del mismo anfitrión.

Las comprobaciones que contradigan estos supuestos se cuantificarán y darán lugar a reglas o limitaciones explícitas, sin modificar silenciosamente la interpretación del estudio.

## Restricciones

La fuente puede presentar omisiones, anuncios desactualizados y diferencias entre las condiciones cotizadas. La extracción es observacional y no corresponde a una muestra probabilística del mercado. Los faltantes y las reglas de selección pueden modificar la composición de los grupos. Además, las ubicaciones son aproximadas y los datos de disponibilidad no distinguen todas las causas de bloqueo del calendario.

La disponibilidad de una sola extracción limita el análisis de cambios históricos y estacionalidad. Las variables de cotización adicionales requieren validación de su contenido porque no están cubiertas por la versión del diccionario oficial consultada. La reconstrucción del entorno depende de la disponibilidad de las versiones declaradas y de la verificación del kernel empleado.

El tiempo disponible para esta entrega limita la incorporación y validación de fuentes adicionales. El trabajo se organiza con las herramientas del curso y se concentra en el archivo seleccionado. Estas condiciones se distinguen de las decisiones de alcance: ciudad, segmentos de comparación y tipo de análisis.

Las estimaciones de ocupación e ingresos del proveedor quedan fuera del análisis principal. Tampoco se realizarán predicciones de rentabilidad ni recomendaciones automáticas de tarifas. Los resultados comunicarán diferencias descriptivas junto con su cobertura y sus limitaciones.

## 1. Organización del proyecto y herramientas

Git registra versiones y permite vincular cada ejecución con una rama y un commit. GitHub aloja la versión compartida; Jupyter integra explicación, código y salidas; pandas permite leer y perfilar las tablas; NumPy se utiliza dentro de las funciones para comprobar valores numéricos finitos. La `.venv` identifica el entorno específico del proyecto.

Se localiza la raíz del repositorio desde la carpeta actual o sus directorios superiores. Las funciones se importan desde `src/`; los datos originales se leen de `data/raw/`; los diagnósticos documentales se guardarán en `docs/evidencias/`. Este bloque no crea datos procesados.

```python
from pathlib import Path
from datetime import datetime, timezone
import sys

from IPython.display import display, Markdown

candidatos = [Path.cwd().resolve(), *Path.cwd().resolve().parents]
RAIZ = next(
    (p for p in candidatos
     if (p / "README.md").is_file()
     and (p / "data" / "raw").is_dir()
     and (p / "src" / "reconocimiento.py").is_file()),
    None,
)
if RAIZ is None:
    raise FileNotFoundError("Abrir el notebook desde la carpeta completa del proyecto; se requieren README.md, data/raw y src/reconocimiento.py.")
if str(RAIZ / "src") not in sys.path:
    sys.path.insert(0, str(RAIZ / "src"))

import reconocimiento as rec
import pandas as pd

RUTA_ORIGINAL = RAIZ / "data" / "raw" / "listings.csv.gz"
RUTA_DICCIONARIO = RAIZ / "docs" / "datos" / "diccionario_variables.csv"
RUTA_EVIDENCIA = RAIZ / "docs" / "evidencias" / "reconocimiento_f1.json"
SHA256_REFERENCIA = "c8db6461c06c0de3efbdda6821d6d28c9aef7dff3c6fde64ac4fd3661e15b78a"
FECHA_EJECUCION_UTC = datetime.now(timezone.utc).isoformat()

estructura = pd.DataFrame([
    {"ruta": ruta, "disponible": (RAIZ / ruta).exists()}
    for ruta in ["F1", "F2", "src", "data/raw", "data/processed", "docs", "README.md", "requirements.txt"]
])
display(estructura)
```

## 2. Verificación del entorno y las dependencias

La comprobación se realiza dentro del kernel. Se registran Python y cada dependencia directa declarada en `requirements.txt`, con importación efectiva, versión instalada y concordancia con las versiones fijadas. `pip check` revisa las relaciones de dependencia instaladas.

La ejecución se detiene si el kernel no pertenece a la `.venv` del proyecto, si no utiliza Python 3.13 o si las dependencias declaradas no cumplen sus versiones. Las entradas sin versión fijada quedan señaladas y requieren conservar el inventario del entorno para el cierre reproducible.

```python
entorno, dependencias = rec.verificar_entorno(RAIZ)
display(entorno)
display(dependencias)

estado_entorno = entorno.set_index("componente")["valor"]
if not bool(estado_entorno["Kernel en .venv del proyecto"]):
    raise RuntimeError("Seleccionar el kernel de la .venv del proyecto y reiniciar la ejecución.")
if not bool(estado_entorno["Entorno virtual activo"]):
    raise RuntimeError("El intérprete no corresponde a un entorno virtual activo.")
if not bool(estado_entorno["Python de la serie 3.13"]):
    raise RuntimeError("El proyecto declara Python 3.13; revisar el kernel seleccionado.")
if not bool(estado_entorno["pip check correcto"]):
    raise RuntimeError("pip check detectó dependencias incompatibles; revisar su salida.")
if not dependencias["cumple_declaracion"].all():
    raise RuntimeError("Hay dependencias ausentes o versiones que no cumplen requirements.txt.")

n_sin_fijar = int((~dependencias["version_fijada"]).sum())
display(Markdown(f"**Entorno verificado.** Dependencias directas sin versión fijada: **{n_sin_fijar}**. Las versiones efectivamente instaladas se conservarán en la evidencia y en el inventario de dependencias."))
```

## 4. Fuente e integridad

La fuente es el archivo detallado de Inside Airbnb para Buenos Aires identificado por la ruta `2026-09-23` (Inside Airbnb, 2026). La descarga local se registró el 22 de noviembre de 2026 a las 23:27:27; esa fecha se distingue de la extracción por anuncio. [Ficha de procedencia](../docs/datos/fuente_dataset.md).

Se calcula el SHA-256 del archivo comprimido antes de leerlo. Debe coincidir con la copia de referencia revisada; el nombre y las dimensiones no bastan para acreditar identidad de contenido. El archivo se lee completo, con las 90 columnas originales.

```python
sha256_inicial = rec.calcular_sha256(RUTA_ORIGINAL)
display(pd.DataFrame([{
    "archivo": str(RUTA_ORIGINAL.relative_to(RAIZ)),
    "bytes": RUTA_ORIGINAL.stat().st_size,
    "sha256": sha256_inicial,
    "coincide_referencia": sha256_inicial == SHA256_REFERENCIA,
}]))
if sha256_inicial != SHA256_REFERENCIA:
    raise ValueError("El archivo difiere de la copia de referencia; revisar procedencia y versión antes de continuar.")

datos = rec.cargar_dataset(RUTA_ORIGINAL)
huella_tabla_inicial = rec.huella_tabla(datos)
resumen = rec.resumir_dataset(datos)
display(resumen)
assert datos.shape == (29753, 90), "Las dimensiones difieren de la fuente documentada."
```

## 5. Reconocimiento de variables

Se miden tipos leídos, faltantes y cardinalidad, y se vinculan con los roles del diccionario manual. Los identificadores se interpretan como códigos; `price` representa un importe aunque su formato sea texto; barrio y tipo de alojamiento son variables nominales. Una columna totalmente vacía no acredita información del rol que conceptualmente tendría.

El diccionario oficial consultado corresponde a la versión 4.3 y no cubre todos los campos adicionales de esta extracción. Las marcas O, U y P distinguen respaldo documental, definiciones incompletas e interpretaciones provisionales, según [los criterios del diccionario](../docs/datos/diccionario_variables.md) (Inside Airbnb, s. f.-b).

```python
perfil = rec.perfilar_variables(datos, RUTA_DICCIONARIO)
with pd.option_context("display.max_rows", 100, "display.max_columns", 10):
    display(perfil)

roles = perfil.groupby("rol_analitico", sort=True).size().rename("n_columnas").reset_index()
display(roles)
assert len(perfil) == datos.shape[1]
assert perfil["variable"].is_unique
```

### Faltantes y observaciones iniciales

El aviso del validador corresponde al archivo original. En este bloque se identifican las columnas que lo originan, sin eliminarlas ni completarlas. La selección de variables y las alternativas para tratar faltantes se justificarán con sus efectos en F2. [Informe del validador](../docs/datos/validacion_dataset.md).

```python
columnas_con_aviso = perfil.loc[
    perfil["porcentaje_faltantes"] > 60,
    ["variable", "rol_analitico", "n_faltantes", "porcentaje_faltantes", "uso_propuesto"],
]
display(columnas_con_aviso)
columnas_vacias = perfil.loc[perfil["n_faltantes"] == len(datos), "variable"].tolist()
display(Markdown(
    f"Se observan **{len(columnas_vacias)} columnas completamente vacías** y "
    f"**{len(columnas_con_aviso)} columnas con más del 60 % de faltantes**. "
    "Estos resultados no implican exclusiones ni imputaciones ejecutadas."
))
```

## 6. Condiciones de las cotizaciones

Se reconoce el contenido de moneda y fechas para fundamentar la pregunta. La lectura temporal de JSON y fechas se realiza en vistas de diagnóstico: no se agregan columnas, no se sustituyen valores y no se produce una base procesada. Los conteos permiten distinguir datos ausentes, estructuras no válidas y condiciones de estancia heterogéneas.

Los importes con símbolo `$` no se equiparan automáticamente con ARS. Las fechas de cotización tampoco representan reservas realizadas. Esta distinción orientará la selección y las comparaciones de F2.

```python
cotizaciones = rec.observar_cotizaciones(datos)
display(cotizaciones["resumen"])
display(cotizaciones["monedas"])
display(cotizaciones["fechas"])

assert sum(cotizaciones["resumen"].iloc[:5]["valor"]) == len(datos)
```

## 7. Comprobación de conservación y evidencia de ejecución

Se vuelve a calcular la huella del archivo y de la tabla leída. La coincidencia confirma que el reconocimiento no cambió la fuente ni la tabla en memoria. Solo entonces se guarda un JSON de documentación con los resultados y las versiones reales de esta ejecución. El archivo de evidencia no es un dataset procesado.

```python
sha256_final = rec.calcular_sha256(RUTA_ORIGINAL)
huella_tabla_final = rec.huella_tabla(datos)
assert sha256_final == sha256_inicial, "El archivo original cambió durante la ejecución."
assert huella_tabla_final == huella_tabla_inicial, "La tabla original cambió durante el reconocimiento."

evidencia = {
    "fase": "F1",
    "fecha_ejecucion_utc": FECHA_EJECUCION_UTC,
    "alcance": "Definición, entorno y reconocimiento inicial; sin limpieza ni imputación",
    "entorno": rec.tabla_a_registros(entorno),
    "dependencias": rec.tabla_a_registros(dependencias),
    "git": git,
    "fuente": {
        "ruta": str(RUTA_ORIGINAL.relative_to(RAIZ)),
        "sha256": sha256_final,
        "coincide_referencia": sha256_final == SHA256_REFERENCIA,
        "archivo_sin_cambios": sha256_final == sha256_inicial,
        "tabla_sin_cambios": huella_tabla_final == huella_tabla_inicial,
    },
    "resumen": rec.tabla_a_registros(resumen),
    "perfil_variables": rec.tabla_a_registros(perfil),
    "cotizaciones": {nombre: rec.tabla_a_registros(tabla) for nombre, tabla in cotizaciones.items()},
}
rec.guardar_evidencia(RUTA_EVIDENCIA, evidencia)
display(Markdown(
    f"**Reconocimiento completado sin modificar los datos.** Evidencia guardada en "
    f"`{RUTA_EVIDENCIA.relative_to(RAIZ).as_posix()}`. Guardar también el notebook con sus salidas."
))
```
